```python
#How to handel null or missing valus
#NAN = Not a number 
#All these strings are considered as NAN valus by pandas
##N/A      -NaN        null
##N/A N/A  -nan        n/a
##NA       N/A         nan
#-1.#IND   NA          1.#IND
#-1.#QNAN  NULL        1.#QNAN
df = pd.read_csv('C:/Users/DELL/Downloads/New.csv',na_values = ['not available','no values'])
df
```
