Mathematic Operation using Numpy


```python
import numpy as np
```

```python
arr1 = np.arange(1,10).reshape(3,3)
arr2 = np.arange(1,10).reshape(3,3)
print(arr1)
print(arr2)
```

```python
arr1+arr2
```

```python
np.add(arr1,arr2)
```

```python
arr1-arr2
```

```python
np.subtract(arr1,arr2)
```

```python
arr1/arr2
```

```python
np.divide(arr1,arr2)
```

```python
arr1 * arr2
```

```python
np.multiply(arr1,arr2)
```

```python
#matrix product
arr1 @ arr2
```

```python
arr1.dot(arr2)
```

```python
arr1
```

```python
#To find the maximum value
arr1.max()
```

```python
#To find the index of the maximum value
arr1.argmax()
```

```python
 #axis = 0 means max element in col and axis = 1 means max element in row
arr1.max(axis = 1)
```

```python
arr1.max(axis = 0)
```

```python
#To find the minimum value in the array
arr1.min()
```

```python
#To find the index of minimum array value
arr1.argmin()
```

```python
arr1.min(axis = 0) #col
```

```python
arr1.min(axis = 1) #row
```

```python
arr1
```

```python
#To find the sum of all elements in the array
np.sum(arr1)
```

```python
#To find the sum of all columns
np.sum(arr1,axis = 0)
```

```python
#To find the sum of all rows in a matrix
np.sum(arr1,axis = 1)
```

```python
#To find the mean of an array
np.mean(arr1)
```

```python
#To find the squareroot of every element
np.sqrt(arr1) 
```

```python
#To find the standard division of an array
np.std(arr1)
```

```python
#To find the exponent of every element
np.exp(arr1)
```

```python
#To find the log of every element
np.log(arr1)
```

```python
#To find log base 10 of every element
np.log10(arr1)
```
