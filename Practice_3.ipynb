#NumPy Functions:

```python
import numpy as np
#arrays can be created using these functions
```

arange()

```python
#np.arange(start,end,steps)
#array with range
ar_1d = np.arange(1,13)
print(ar_1d)
```

```python
even_array = np.arange(2,13,2)
print(even_array)
```

linspace()

```python
#Same difference in all array values with in the given range
np.linspace(1,5,4)
```

reshape()

```python
#Change the array in different combination multidimentional array
ar_2d = ar_1d.reshape(3,4)
print(ar_2d)
```

```python
ar_2d = ar_1d.reshape(2,3,2)
print(ar_2d)
```

```python
ar = np.arange(1,13).reshape(2,6)
print(ar)
```

ravel()

```python
#converts multi-dimensional array to 1D array
ar.ravel()
```

flatten()

```python
#converts multi-dimensional array to 1D array
ar.flatten()
#It can take an argument i.e order = 'C'......(Learn)
```

transpose()

```python
#Conerts row into column and column into row
ar.transpose()
```

```python
#transpose() can also be written as .T
ar.T
```
