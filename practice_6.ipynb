Python Numpy Array Concatination and split


```python
import numpy as np
```

```python
arr1 = np.arange(1,17).reshape(4,4)
print(arr1)
```

```python
arr2 = np.arange(17,33).reshape(4,4)
print(arr2)
```

```python
list1 = [2,4,5,6]
list2 = [1,2,3,4]
list1 + list2
```

```python
arr1 + arr2 #normal addition
```

```python
np.concatenate((arr1,arr2))#Concatenation
```

```python
np.concatenate((arr1,arr2),axis = 1)#Row wise concatenation
```

```python
np.vstack((arr1,arr2))#vertical concatenation
```

```python
np.hstack((arr1,arr2))#horizontal concatenation
```

```python
np.hstack((arr1,arr2,arr1))
```

```python
np.split(arr1,2) #splits the array and store the parts(multiple arrays) in a list
```

```python
list1 = np.split(arr1,2)
type(list1)
```

```python
list1[0]
```

```python
type(list1[0])
```

```python
np.split(arr1,2,axis = 1)
```

```python
d1 = np.array([4,7,1,3,9])
```

```python
#splitting it into multiple array
np.split(d1,[1,3])
```
