How To Create Python Pandas Dataframe


```python
import pandas as pd
```

```python
#Creating an empty dataframe
#DataFrame method is used
emt_df = pd.DataFrame()
print(emt_df)
```

```python
lst = ['a','b','c']
print(lst)
```

```python
df1 = pd.DataFrame(lst)
print(df1)
```

```python
df1
```

```python
ls_of_ls = [[1,2,3],[2,3,4],[4,5,6]]
print(ls_of_ls)
```

```python
df2 = pd.DataFrame(ls_of_ls)
df2
```

```python
dict1 = {'ID':[11,22,33,44]}
dict1
```

```python
df3 = pd.DataFrame(dict1)
df3
```

```python
#ID is the label
dict1 = {'ID':[11,22,33,44],'SN':[1,2,3,4]} #The value of each data should be same

dict1
```

```python
df3 = pd.DataFrame(dict1)
df3
```

```python
ls_dict = [{'a':1,'b':2},{'a':3,'b':4}] #List of dictionaries
df5 = pd.DataFrame(ls_dict)
df5
```

```python
ls_dict = [{'a':1,'b':2},{'a':3,'b':4,'c':5}] #List of dictionaries
df5 = pd.DataFrame(ls_dict)
df5
```

```python
dict_sr = {'ID':pd.Series([1,2,3]), 'SN' : [111,222,333]}
df6 = pd.DataFrame(dict_sr)
df6
```
