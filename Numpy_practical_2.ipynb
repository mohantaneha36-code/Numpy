```python
import numpy as np
```

```python
mx_1s = np.array([[1,1,1,],[1,1,1],[1,1,1]])
print(mx_1s)
```

```python
#matrix
```

```python
mx_1s = np.ones(5)
print(mx_1s)
#ones((row,column)) used to create multiple rows and columns
```

```python
mx_1s.dtype
#ones() stores float values
```

```python
mx_1s = np.ones((3,4))
print(mx_1s)
```

```python
#changing the type of data to int
#np.ones(shape,dtype=None,order='c')
mx_1s = np.ones((3,4),dtype=int)
print(mx_1s)
```

```python
#To create a matrix with only zeros zeros((row,col),dtype) is used
mx_0s = np.zeros((4,6))
print(mx_0s)
```

```python
mx_0s = np.zeros((4,6),dtype=bool)
print(mx_0s)
```

```python
mx_0s = np.zeros((4,6),dtype=str)
print(mx_0s)
#empty strings are printed
```

```python
em_str = ''
print(bool(em_str))
```

```python
#if empty string is converted to bool result will be false
em_mx = np.empty((3,3))
print(em_mx)
#empty() matrix can contain any random value
```
