Python Numpy Array Slicing(:) 

```python
import numpy as np
```

```python
mx = np.arange(1,101).reshape(10,10)
print(mx) #consider it as an imaginary file
```

```python
#To get an element from the matrix
mx[0,0]
```

```python
#To check the dimension
mx[0,0].ndim
#This value doesnot contain any dimension i.e it is a scalar value
```

```python
#To get  single row and all columns
mx[0]
```

```python
#To get single column
mx[:,0]
```

```python
#To get a 2d array of the above array using slice operator
mx[:,0:2]
```

```python
mx[:,0:2].ndim
```

```python
mx
```

```python
#To get 2nd row ,3rd row,4th row
mx[1:4,1:4]
```

```python
mx[:,1:3]
```

```python
mx[:]
```

```python
mx[::]
```

```python
mx[:,:]
```

```python
#Space required to store the matrix element
mx.itemsize
```
