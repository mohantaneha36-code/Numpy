```python
#Series is a type of data structure that we use in pandas
#Creting series using list
#In a list different data types can be stored
list_s = [1,2,-3,6.2,'data values']
print(list_s)
```

```python
series1 = pd.Series(list_s) #To create a series in pandas we use the Series() method
print(series1)
#first column is for index values and the second column is for data values
#dtype : object bcz the list stores different different types of data
```

```python
type(series1)
```

```python
#Series is a one dimensional indexed array
series2 = pd.Series([1,2,3,4])
print(series2)
```

```python
empty_s = pd.Series([]) #Empty series
print(empty_s)
```

```python
series3 = pd.Series([1,2,3,4], index = ['a','b','c','d']) #Changing the index
print(series3)
```

```python
series3 = pd.Series([1,2,3,4], index = ['a','b','c','d'] , dtype = float) #Changing the dtype
print(series3)
```

```python
series3 = pd.Series([1,2,3,4], index = ['a','b','c','d'], dtype = float , name = 'data values') #Changing the index
print(series3)
```

```python
scalar_s = pd.Series(0.5)
print(scalar_s)
```

```python
scalar_s = pd.Series(0.5, index = [1,2,3])
print(scalar_s)
```

```python
dict_s = pd.Series({'a':1,'b':2})
print(dict_s)
```
